6. Rectangle Area
   Create variables for the length and width of a rectangle. Calculate and print its area and 
perimeter.

```python
rectangle_length=10
rectangle_width=5
rectangle_area=rectangle_length*rectangle_width
perimeter=2*(rectangle_length+rectangle_width)
print("Area of rectangle:",rectangle_area)
print("Perimeter is:",perimeter)
```

7. Time Converter
   Store 250 minutes in a variable. Calculate and print the number of hours and remaining 
minutes.

```python
time_in_minutes=250
convert_in_hour=time_in_minutes//60
remaining_minutes=time_in_minutes%60
print(f"250 minutes has {convert_in_hour} hours and {remaining_minutes} minutes")
```

8. Student Eligibility
   Create variables for marks and attendance. 
A student is eligible if: 
• Marks are greater than or equal to 50  
• Attendance is greater than or equal to 75
