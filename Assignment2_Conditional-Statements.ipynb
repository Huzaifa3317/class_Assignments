Question 8: ATM Withdrawal 
You are developing an ATM withdrawal system. 
The user's account balance is: 
50000 
Ask the user to enter the withdrawal amount. 
Apply these rules: 
• If the amount is less than or equal to 0, display "Invalid Amount".  
• If the amount is greater than the balance, display "Insufficient Balance".  
• Otherwise, process the withdrawal and display the remaining balance. 

```python
user_account_balance=50000
withdrawal_amount=int(input("Enter Withdrawal Amount"))
if withdrawal_amount<=0:
    print("Invalid Amount")
elif withdrawal_amount>user_account_balance:
    print("Insufficient Balance")
else:
    remaining_balance=user_account_balance-withdrawal_amount
    print(f"Withdrawal Successful and Your Remaining Balance is {remaining_balance}")
```

Question 12: Employee Performance & Promotion System 
A company wants to automatically evaluate its employees for promotion. 
Ask the employee for: 
• Employee name  
• Years of experience  
• Monthly salary  
• Performance score out of 100  
• Number of projects completed  
• Number of late days  
Promotion Rules 
An employee can be promoted only if: 
• Experience is at least 2 years.  
• Performance score is at least 70.  
• At least 3 projects have been completed.  
• Late days are less than 10.  
If any requirement is not satisfied, the employee is Not Eligible for Promotion. 
The program should display the specific reason. 
Promotion Level 
If the employee is eligible: 
Performance >= 90 
AND Experience >= 5 
→ Senior Promotion 
Performance >= 80 
AND Experience >= 3 
→ Mid-Level Promotion 
Otherwise 
→ Junior Promotion 
Bonus 
Calculate the bonus category: 
Performance >= 90 
→ 20% bonus 
Performance >= 80 
→ 15% bonus 
Performance >= 70 
→ 10% bonus 
Otherwise 
→ No bonus 
Finally, display: 
===== EMPLOYEE REPORT ===== 
Employee: Ahmed 
Experience: 4 years 
Performance: 87 
Projects: 6 
Late Days: 3 
Promotion Status: Eligible 
Promotion Level: Mid-Level 
Bonus: 15% 

```python
reason = ""

employee_name = input("Enter Employee Name: ")
employee_experience = int(input("Enter Years of Experience: "))
employee_salary = int(input("Enter Monthly Salary: "))
employee_performance = int(input("Enter Performance Score out of 100: "))
employee_projects = int(input("Enter Number of Projects Completed: "))
employee_late_days = int(input("Enter Number of Late Days: "))

if (employee_experience >= 2
    and employee_performance >= 70
    and employee_projects >= 3
    and employee_late_days < 10):

    status = "Eligible"

    if employee_performance >= 90 and employee_experience >= 5:
        promotion = "Senior Promotion"
    elif employee_performance >= 80 and employee_experience >= 3:
        promotion = "Mid-Level Promotion"
    else:
        promotion = "Junior Promotion"

    if employee_performance >= 90:
        bonus = "20%"
    elif employee_performance >= 80:
        bonus = "15%"
    elif employee_performance >= 70:
        bonus = "10%"
    else:
        bonus = "No Bonus"

else:
    status = "Not Eligible"
    promotion = "N/A"
    bonus = "No Bonus"

    if employee_experience < 2:
        reason += "Experience must be at least 2 years. "

    if employee_performance < 70:
        reason += "Performance score must be at least 70. "

    if employee_projects < 3:
        reason += "At least 3 projects must be completed. "

    if employee_late_days >= 10:
        reason += "Late days must be less than 10. "

print("===== EMPLOYEE REPORT =====")
print(f"Employee: {employee_name}")
print(f"Experience: {employee_experience} years")
print(f"Monthly Salary: {employee_salary}")
print(f"Performance: {employee_performance}")
print(f"Projects: {employee_projects}")
print(f"Late Days: {employee_late_days}")
print(f"Promotion Status: {status}")

if status == "Not Eligible":
    print(f"Reason: {reason}")
else:
    print(f"Promotion Level: {promotion}")
    print(f"Bonus: {bonus}")
```
