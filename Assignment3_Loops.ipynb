Question 7: ATM Menu 
Create a simple ATM system using a while loop. 
The initial balance is: 
50000 
Display this menu repeatedly: 
1. Check Balance 
2. Deposit 
3. Withdraw 
4. Exit 
The program should continue until the user selects 4. 
Implement the following: 
• Check balance  
• Deposit money  
• Withdraw money  
• Prevent withdrawal if the amount is greater than the balance  
• Exit the program  

```python
balance = 50000

while True:
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        print(f"Your Balance: Rs. {balance}")

    elif choice == 2:
        amount = float(input("Enter deposit amount: "))
        balance += amount
        print(f"Amount deposited. New Balance: Rs. {balance}")

    elif choice == 3:
        amount = float(input("Enter withdrawal amount: "))

        if amount > balance:
            print("Insufficient Balance")
        else:
            balance -= amount
            print(f"Withdrawal successful. New Balance: Rs. {balance}")

    elif choice == 4:
        print("Thank you for using ATM.")
        break

    else:
        print("Invalid choice. Try again.")
```

Question 8: Number Guessing Game 
Create a number guessing game. 
Set the secret number to: 
secret_number = 25 
Ask the user to guess the number. 
Use a while loop. 
If the guess is: 
• Greater than 25 → "Too High"  
• Less than 25 → "Too Low"  
• Equal to 25 → "Correct!"  
Also count how many attempts the user made. 
Example: 
Guess: 30 
Too High 
Guess: 20 
Too Low 
Guess: 25 
Correct! 
Attempts: 3 

```python
secret_number = 25
attempts = 0

while True:
    guess = int(input("Guess: "))
    attempts += 1

    if guess > secret_number:
        print("Too High")
    elif guess < secret_number:
        print("Too Low")
    else:
        print("Correct!")
        break

print(f"Attempts: {attempts}")
```

Question 9: Student Marks Entry System 
A teacher wants to enter student marks continuously. 
Use a while loop. 
The program should keep asking for marks until the teacher enters: -1 
For every student's marks: 
• 50 or above → Pass  
• Below 50 → Fail  
At the end, display: 
Total Students 
Total Passed 
Total Failed 
The -1 should not be counted as a student's marks. 

```python
students = 0
passed = 0
failed = 0

marks = int(input("Enter student marks (-1 to stop): "))

while marks != -1:
    students += 1

    if marks >= 50:
        print("Pass")
        passed += 1
    else:
        print("Fail")
        failed += 1

    marks = int(input("Enter student marks (-1 to stop): "))

print(f"Total Students: {students}")
print(f"Total Passed: {passed}")
print(f"Total Failed: {failed}")
```

Question 11: Student Performance Management System 
A teacher wants to create a system for analyzing the performance of multiple students. 
First ask: 
Enter number of students: 
For each student, ask: 
• Student name  
• Marks in Python  
• Marks in Mathematics  
• Marks in English  
Calculate: 
Total Marks 
Percentage 
Result Rules 
A student is considered Pass only if: 
• Overall percentage is at least 50%  
• AND marks in every subject are at least 40.  
Otherwise, the student is Fail. 
Grade Rules 
For students who pass: 
90+     → A+ 
80-89   → A 
70-79   → B 
60-69   → C 
50-59   → D 
While processing students, calculate: 
• Number of passed students  
• Number of failed students  
• Highest percentage  
• Lowest percentage  
• Class total marks  
• Class average percentage  
Final Display: 
===== CLASS REPORT ===== 
Total Students: 10 
Passed: 7 
Failed: 3 
Highest Percentage: 94% 
Lowest Percentage: 42% 
Class Average: 76.5% 
Important requirement 
Use a loop to process all students. 
Do not manually write the same code for every student.

```python
students = int(input("Enter number of students: "))

passed = 0
failed = 0
highest = 0
lowest = 100
class_total = 0
class_percentage = 0

for i in range(students):
    name = input("Enter student name: ")
    python = float(input("Enter Python marks: "))
    maths = float(input("Enter Mathematics marks: "))
    english = float(input("Enter English marks: "))

    total = python + maths + english
    percentage = (total / 300) * 100

    print(f"\nStudent Name: {name}")
    print(f"Total Marks: {total}")
    print(f"Percentage: {percentage}%")

    if percentage >= 50 and python >= 40 and maths >= 40 and english >= 40:
        print("Result: Pass")
        passed += 1

        if percentage >= 90:
            print("Grade: A+")
        elif percentage >= 80:
            print("Grade: A")
        elif percentage >= 70:
            print("Grade: B")
        elif percentage >= 60:
            print("Grade: C")
        else:
            print("Grade: D")
    else:
        print("Result: Fail")
        failed += 1

    if percentage > highest:
        highest = percentage

    if percentage < lowest:
        lowest = percentage

    class_total += total
    class_percentage += percentage

average = class_percentage / students

print("\n===== CLASS REPORT =====")
print(f"Total Students: {students}")
print(f"Passed: {passed}")
print(f"Failed: {failed}")
print(f"Highest Percentage: {highest}%")
print(f"Lowest Percentage: {lowest}%")
print(f"Class Total Marks: {class_total}")
print(f"Class Average: {average}%")
```

Question 12: Restaurant Ordering & Billing System 
You are developing a restaurant ordering system. 
The restaurant has the following menu: 
MENU  
1. Burger       
2. Pizza        
3. Biryani      
4. Fries        
5. Drink        
6. Checkout 
Rs. 500 
Rs. 1000 
Rs. 350 
Rs. 250 
Rs. 150 
The program should repeatedly display the menu and allow the customer to order multiple 
items. 
The program should continue until the customer selects: 
6. Checkout 
When an item is selected 
Add its price to the total bill. 
Also keep track of how many items the customer ordered. 
For example: 
Burger added 
Pizza added 
Drink added 
Invalid Input 
If the customer enters something other than 1-6: 
Invalid choice. Please try again. 
The program should not terminate. 
Discount Rules 
When the customer checks out: 
Bill >= 5000 
→ 20% discount 
Bill >= 3000 
→ 15% discount 
Bill >= 1500 
→ 10% discount 
Below 1500 
→ No discount 
Delivery Charges 
After discount: 
Final bill >= 3000 
→ Free Delivery 
Final bill < 3000 
→ Rs. 200 Delivery Charges 
Finally display: 
===== FINAL BILL ===== 
Total Items: 5 
Original Bill: Rs. 3500 
Discount: Rs. 525 
Delivery Charges: Rs. 0 
Final Bill: Rs. 2975 
Thank you for ordering! 

```python
total = 0
items = 0

while True:
    print("1. Burger - Rs. 500")
    print("2. Pizza - Rs. 1000")
    print("3. Biryani - Rs. 350")
    print("4. Fries - Rs. 250")
    print("5. Drink - Rs. 150")
    print("6. Checkout")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        total += 500
        items += 1
        print("Burger added")

    elif choice == 2:
        total += 1000
        items += 1
        print("Pizza added")

    elif choice == 3:
        total += 350
        items += 1
        print("Biryani added")

    elif choice == 4:
        total += 250
        items += 1
        print("Fries added")

    elif choice == 5:
        total += 150
        items += 1
        print("Drink added")

    elif choice == 6:
        break

    else:
        print("Invalid choice. Please try again.")

if total >= 5000:
    discount = total * 0.20
elif total >= 3000:
    discount = total * 0.15
elif total >= 1500:
    discount = total * 0.10
else:
    discount = 0

bill = total - discount

if bill >= 3000:
    delivery = 0
else:
    delivery = 200

final_bill = bill + delivery

print("===== FINAL BILL =====")
print(f"Total Items: {items}")
print(f"Original Bill: Rs. {total}")
print(f"Discount: Rs. {discount}")
print(f"Delivery Charges: Rs. {delivery}")
print(f"Final Bill: Rs. {final_bill}")
print("Thank you for ordering!")
```
